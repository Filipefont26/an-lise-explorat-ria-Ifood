```python
# número total de clientes

total_clientes = df['Index'].nunique()
print(f"numero total de clientes : ",total_clientes)
```

```python
#receita total
receita_total = df['MntTotal'].sum()
print(f"Receita Total : R$ {receita_total:,.2f}")
```

```python
#Ticket Médio

Ticket_medio = df['MntTotal'].mean()
print(f"Ticket Médio :", Ticket_medio)
```

```python
#taxa de conversão

df ['comprou'] = (

df ['NumDealsPurchases']+
df ['NumWebPurchases']+
df ['NumCatalogPurchases']  +
df ['NumStorePurchases']
) > 0

taxa_coversao = df['comprou'].mean() * 100
print(f"taxa de conversão : {taxa_coversao:.2f}%")
```

```python
#  distribuição de compras por canal

df['total_compras'] =(
df ['NumDealsPurchases']+
df ['NumWebPurchases']+
df ['NumCatalogPurchases']  +
df ['NumStorePurchases']
)

canais = {
     "web " : int(df['NumWebPurchases'].sum()),
     "catalogo" :int(df['NumCatalogPurchases'].sum()),
     "loja": int(df['NumStorePurchases'].sum()),
     "online" : int(df['NumDealsPurchases'].sum())
}

print(f"compra por canais :{canais}")
```

```python
# Qual categoria de produto mais vende

mix_produtos = df[
    ['MntMeatProducts','MntFishProducts','MntSweetProducts','MntRegularProds','MntGoldProds','MntWines','MntFruits' ]
].sum()

ordenado = mix_produtos.sort_values(ascending=False)

for produto, valor in ordenado.items():
    print(f"{produto}: R${valor:,.2f}")
```

```python
# ANÁLISE DE CLIENTES
print("Detalhes de idade de clientes : ")
df['Age'].describe()
```

```python
# classificação e gasto por idade

df['faixa_etaria'] = pd.cut(
    df['Age'],
    bins=[18,30,40,50,60,100],
    labels=['18-30','31-40','41-50','51-60','60+']

)

df['faixa_etaria'].value_counts()
gasto_por_idade = df.groupby('faixa_etaria')['MntTotal'].mean().sort_values(ascending=False)
print("gasto médio por faixa etária")
print(gasto_por_idade)
```

```python
gasto_por_estado_civil = df.groupby('marital_status')['MntTotal'].sum().sort_values(ascending=False)

print("gasto total por estado civil : ")

for estado, valor in gasto_por_estado_civil.items():
    print(f"{estado}: R${valor:,.2f}")

```

```python
#gasto por renda
renda_gasto=df['Income'].describe().sort_values(ascending=False)

print("Gasto por renda :")

for renda, valor in renda_gasto.items():
    print(f"{renda}: R${valor:,.2f}")



```

```python
#Gasto por educação

gasto_por_educacao= df.groupby('education')['MntTotal'].sum().sort_values(ascending=False)

print("gasto por educação :")

for educação, valor in gasto_por_educacao.items():
    print(f"{educação}: R${valor:,.2f}")
```

```python
#Clientes Alto Valor vs Baixo Valor

df['segmentoValor'] = pd.qcut(df['MntTotal'],4,
       labels=['baixo','medio', 'alto', 'muito alto']                       
                              
)

valores_cliente = df.groupby('segmentoValor')['MntTotal'].mean().sort_values(ascending=False)

print("Clientes Alto Valor vs Baixo Valor :")

for cliente, valor in valores_cliente.items():
    print(f"{cliente}: R${valor:,.2f}")


```

```python
#gasto por filhos


def classificacao_famlia(row):
    if row['Kidhome'] > 0 and row['Teenhome'] > 0:
        return 'criança e adolecente'
    elif row['Kidhome'] > 0:
        return 'apenas criança'
    
    elif row['Teenhome'] > 0:
        return 'apenas adolescente'
    else:
        return 'sem filhos'
    
df['GrupoFilhos'] = df.apply(classificacao_famlia,axis=1)

print("Gasto por filhos")

gasto_por_filho = df.groupby('GrupoFilhos')['MntTotal'].sum().sort_values(ascending=False)

for filhos, valor in gasto_por_filho.items():
    print(f"{filhos}: R${valor:,.2f}")

    
```

```python
#análise de campanhas de marketing

campanhas = df[['AcceptedCmp1','AcceptedCmp2','AcceptedCmp3','AcceptedCmp4','AcceptedCmp5']].sum()
print(campanhas)
```

```python
#taxa de aceitação por campanha

taxa = df[['AcceptedCmp1','AcceptedCmp2','AcceptedCmp3','AcceptedCmp4','AcceptedCmp5']].mean()*100

print("porcentagem por campanha :")

for porcentagem, valor in taxa.items(): 
    print(f"{porcentagem}: {valor:,.2f}%")
```

```python
#usuarios que aceitaram mais campanhas

df['TotalCampanhasAceitas'] = (
    df['AcceptedCmp1'] +
    df['AcceptedCmp2'] +
    df['AcceptedCmp3'] +
    df['AcceptedCmp4'] +
    df['AcceptedCmp5']
)

df['TotalCampanhasAceitas'].value_counts()
```

```python
#Quem aceita as campanhas gasta mais ? 

campanhas_aceitas = df.groupby('TotalCampanhasAceitas')['MntTotal'].mean()

print('Quem aceita as campanhas gasta mais ? ')

for aceitas, valor in campanhas_aceitas.items(): 
    print(f"{aceitas}: {valor:,.2f}")
```

```python
df['TotalCampanhasAceitas'].max()
```

```python
# campanhas atraem clientes de maior valor ou menor valor?

df['AceitouCampanhas'] = df['TotalCampanhasAceitas'] > 0

print('campanhas atraem clientes de maior valor ou menor valor?')

camp_aceita = df.groupby('AceitouCampanhas')['MntTotal'].mean()

for camp, valor in camp_aceita.items(): 
    print(f"{camp}: {valor:,.2f}")

#clientes que respondem a campanhas gastam mais

```
